```python
# Question 1
# Even or Odd
num=int(input("Enter your num:"))

if num % 2 == 0:
  print("Even")

else:
  print("Odd")
```

```python
# Question 2: Simple Calculator
# Take two numbers and an operator (+, -, *, /) from the user.
# Perform the selected operation and display the result.


num1=int(input("Enter first num:"))
num2=int(input ("Enter second num:"))
operator=input("Enter your operator:")
if operator == "/":
  print(num1 / num2)
elif operator == "+":
  print(num1 + num2)
elif operator == "-":
  print(num1 - num2)
elif operator == "*":
  print(num1 * num2)
```

```python
# Question 3: Student Marks
# Take marks of 3 subjects from the user.
# Calculate the total and percentage, then display Pass if the percentage is 50 or above, otherwise display Fail.


subject1 = int(input("Enter first subject marks:"))
subject2 = int(input("Enter second subject marks:"))
subject3 = int(input("Enter third subject marks:"))


total = subject1 + subject2 + subject3
percentage = total/300 * 100

print(total)
print(percentage)

if percentage >= 50:
  print("Pass")

else:
  print("Fail")
```
